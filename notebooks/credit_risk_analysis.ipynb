# Credit Risk Analysis

Análise exploratória da base de inadimplência de cartão de crédito.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

PROJECT_ROOT = Path.cwd().parent
DATA_FILE = PROJECT_ROOT / 'data' / 'processed' / 'credit_default_processed.csv'
df = pd.read_csv(DATA_FILE)
df.head()
```

```python
print('Dimensões:', df.shape)
print('\nValores ausentes:', df.isna().sum().sum())
print('\nDistribuição do target:')
print(df['default'].value_counts())
print('\nProporção do target:')
print(df['default'].value_counts(normalize=True))
```

```python
df.describe().T
```

```python
plt.figure(figsize=(6, 4))
sns.countplot(data=df, x='default')
plt.title('Distribuição de Inadimplência')
plt.xlabel('Inadimplência')
plt.ylabel('Quantidade de clientes')
plt.show()
```

```python
# Comparação do limite de crédito entre clientes adimplentes e inadimplentes
df.groupby('default')['X1'].agg(['count', 'mean', 'median', 'std'])
```
